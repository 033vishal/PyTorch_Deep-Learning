## Tensor Manipulation Operations

```python
# Create a sample tensor
a = torch.arange(12).reshape(3, 4)
print(f'Original tensor a:\n{a}')
```

### `view()` and `reshape()`

- **`view()`**: Returns a new tensor with the same data as `self` tensor but with a different shape. The returned tensor shares the same underlying data with `self`.
- **`reshape()`**: Returns a tensor with the same data and number of elements as `self` but with the specified shape. When possible, the returned tensor will be a view of `self`. Otherwise, it will be a copy.

Both are used for changing the shape of a tensor, but `view` requires the new shape to be compatible with the original tensor's memory layout (i.e., contiguous), while `reshape` can handle non-contiguous tensors by making a copy if necessary.

```python
# Using view - requires contiguous memory
b = a.view(2, 6)
print(f'Tensor after view(2, 6):\n{b}')

# Modifying b also modifies a because they share memory
b[0, 0] = 99
print(f'Original tensor a after modifying b:\n{a}')

# Using reshape - more flexible, can return a copy if needed
c = torch.arange(1, 10).reshape(3, 3)
print(f'Original tensor c:\n{c}')

d = c.t().reshape(3, 3) # .t() makes it non-contiguous
print(f'Tensor after transpose and reshape:\n{d}')

# Modifying d does not modify c because reshape returned a copy
d[0, 0] = 88
print(f'Original tensor c after modifying d:\n{c}')
print(f'Tensor d after modifying:\n{d}')
```

### `squeeze()` and `unsqueeze()`

- **`squeeze()`**: Returns a tensor with all the dimensions of `input` of size 1 removed. If `dim` is given, a squeeze operation is done only in the given dimension.
- **`unsqueeze()`**: Returns a new tensor with a dimension of size one inserted at the specified position. The returned tensor shares the same underlying data with the original tensor.

```python
# Create a tensor with dimensions of size 1
e = torch.zeros(1, 5, 1, 3, 1)
print(f'Original tensor e shape: {e.shape}')

# Squeeze all dimensions of size 1
f = e.squeeze()
print(f'Tensor after squeeze() shape: {f.shape}')

# Squeeze a specific dimension of size 1
g = e.squeeze(2)
print(f'Tensor after squeeze(2) shape: {g.shape}')

# Create a 1D tensor
h = torch.tensor([1, 2, 3])
print(f'Original tensor h shape: {h.shape}')

# Unsqueeze a dimension at position 0
i = h.unsqueeze(0)
print(f'Tensor after unsqueeze(0) shape: {i.shape}')

# Unsqueeze a dimension at position 1
j = h.unsqueeze(1)
print(f'Tensor after unsqueeze(1) shape: {j.shape}')
```

### `flatten()`

- **`flatten()`**: Flattens a contiguous range of dims in a tensor. This function returns a view of the original tensor.

```python
# Create a 3D tensor
k = torch.arange(1, 28).reshape(3, 3, 3)
print(f'Original tensor k:\n{k}\nShape: {k.shape}')

# Flatten the tensor
l = k.flatten()
print(f'Tensor after flatten():\n{l}\nShape: {l.shape}')

# Flatten starting from a specific dimension
m = k.flatten(start_dim=1)
print(f'Tensor after flatten(start_dim=1):\n{m}\nShape: {m.shape}')
```
